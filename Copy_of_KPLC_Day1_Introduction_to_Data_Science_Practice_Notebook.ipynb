## Section 1 — Variables

A **variable** is a named container that stores a value.

Think of it like a labelled box — you put something inside, give the box a name, and use that name later.

In Python, you create a variable like this:

```python
region = 'Nairobi'
```

The name on the left (`region`) is the variable.
The value on the right (`'Nairobi'`) is what you are storing.

Python has four basic types of values you will use every day:

| Type | Name | Example |
|------|------|---------|
| Text | `str` (string) | `'Nairobi'` |
| Whole number | `int` (integer) | `47` |
| Decimal number | `float` | `347.5` |
| Yes / No | `bool` (boolean) | `True` or `False` |


### Guided Example 1.1 — Storing text (a string)

## Section 7 — Introduction to NumPy

**NumPy** (Numerical Python) is a library — a collection of ready-made tools for working with numbers.

You import it once at the top of your notebook, then use its functions anywhere:

```python
import numpy as np
```

The `np` part is just a short nickname so you don't have to type `numpy` every time.

NumPy is designed to work on **lists of numbers** — like a column of meter readings
or a series of monthly bills. Instead of writing a loop to find the average yourself,
you just call `np.mean()` and NumPy does it in one line.

> **Why does Amina need NumPy?** When she has 11,000 billing records, she can't
> calculate statistics by hand. NumPy lets her find the average bill, the highest
> consumption, and flag missing readings — all in a single line of code.


### Guided Example 7.1 — Import NumPy and create an array

An **array** is NumPy's version of a list — but built for maths.
You create one by passing a regular Python list into `np.array()`.


````python
# Import NumPy
import numpy as np

# Monthly kWh readings for one Nairobi customer (12 months)
monthly_units = np.array([245.0, 310.0, 289.5, 401.0, 365.0,
                           512.0, 488.0, 320.0, 275.0, 349.5,
                           410.0, 395.0])

print(monthly_units)
print(type(monthly_units))
````

### Guided Example 7.2 — Basic NumPy functions

NumPy gives you ready-made functions for common calculations.
Here are the ones you will use most often:

| Function | What it does |
|----------|--------------|
| `np.mean()` | Average (sum ÷ count) |
| `np.median()` | Middle value when sorted |
| `np.max()` | Largest value |
| `np.min()` | Smallest value |
| `np.sum()` | Total of all values |
| `np.std()` | Standard deviation (how spread out the values are) |
| `np.round()` | Round to a number of decimal places |
| `np.sqrt()` | Square root |
| `np.nan` | Represents a missing value |
| `np.nanmean()` | Average that skips missing values |


````python
# Mean — average monthly consumption
avg = np.mean(monthly_units)
print(f'Average monthly consumption: {avg} kWh')
````

````python
# Median — the middle value
med = np.median(monthly_units)
print(f'Median monthly consumption: {med} kWh')
````

````python
# Max and Min — highest and lowest readings
print(f'Highest month: {np.max(monthly_units)} kWh')
print(f'Lowest month : {np.min(monthly_units)} kWh')
````

````python
# Sum — total units for the year
total = np.sum(monthly_units)
print(f'Total units for the year: {total} kWh')
````

````python
# Std — how much consumption varies month to month
spread = np.std(monthly_units)
print(f'Standard deviation: {np.round(spread, 2)} kWh')
````

````python
# Round — tidy up a result to 1 decimal places
avg_rounded = np.round(np.mean(monthly_units), 1)
print(f'Average (rounded): {avg_rounded} kWh')
````

````python
# np.nan — how Python represents a missing value
# Some months may have no reading (meter fault, etc.)
readings_with_gap = np.array([245.0, np.nan, 310.0, np.nan, 289.5])

# np.nanmean() calculates the average but skips the missing values
safe_avg = np.nanmean(readings_with_gap)
print(f'Average (ignoring missing): {safe_avg} kWh')
````
